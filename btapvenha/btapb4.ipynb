```python
import os

# --- CÂU 1: Hàm bổ trợ phân tách dữ liệu ---
def parse_city_coordinates(line):
    line = line.strip()
    if not line:  # Bỏ qua dòng trống
        return None
    try:
        parts = line.split(",")
        if len(parts) < 3:
            raise ValueError("Thiếu dữ liệu (Cần: Tên, Vĩ độ, Kinh độ)")
        
        name = parts[0].strip()
        lat = float(parts[1])
        lon = float(parts[2])
        return name, lat, lon
    except ValueError as e:
        print(f"⚠️ Lỗi định dạng tại dòng '{line}': {e}")
        return None

# --- CÂU 1 (Tiếp theo): Hàm đọc tập tin ---
def read_cities(input_file):
    if not os.path.exists(input_file):
        print(f"❌ Lỗi: Không tìm thấy tập tin '{input_file}' tại {os.getcwd()}")
        return None # Trả về None để phân biệt với file trống []
    
    city_list = []
    try:
        with open(input_file, "r", encoding="utf-8") as infile:
            for line in infile:
                data = parse_city_coordinates(line)
                if data:
                    city_list.append(data)
        return city_list
    except Exception as e:
        print(f"❌ Lỗi không xác định khi đọc file: {e}")
        return None

# --- CÂU 2: Hàm ghi danh sách tọa độ ---
def write_city_coordinates(output_file, city_data):
    try:
        with open(output_file, "w", encoding="utf-8") as outfile:
            for name, lat, lon in city_data:
                outfile.write(f"City: {name}, Latitude: {lat}, Longitude: {lon}\n")
        print(f"✅ Đã ghi {len(city_data)} thành phố vào file '{output_file}'")
    except Exception as e:
        print(f"❌ Không thể ghi file output: {e}")

# --- CÂU 3: Hàm xử lý tổng hợp ---
def process_geospatial_file(input_file, output_file):
    # 1. Đọc dữ liệu
    cities = read_cities(input_file)
    
    # 2. Kiểm tra nếu có dữ liệu mới ghi file
    if cities: # Nếu cities có dữ liệu (không phải None và không rỗng)
        write_city_coordinates(output_file, cities)
        print(f"🏁 Hoàn tất xử lý tập tin {input_file}")
    elif cities == []:
        print(f"⚠️ Tập tin {input_file} trống, không có gì để xử lý.")
    else:
        # Trường hợp cities là None (lỗi FileNotFoundError đã in ở trên)
        print("⏭️ Đã dừng xử lý do lỗi đầu vào.")

# --- CHƯƠNG TRÌNH CHÍNH ---
if __name__ == "__main__":
    # Tên file - Đảm bảo file này nằm cùng thư mục với file .py của bạn
    in_f = "coordinates.txt"
    out_f = "output_coordinates.txt"
    
    # (Tùy chọn) Tạo file mẫu nếu bạn muốn chạy thử ngay
    # with open(in_f, "w") as f: f.write("Hanoi, 21.0285, 105.8542\nSaigon, 10.8231, 106.6297")

    process_geospatial_file(in_f, out_f)
```
