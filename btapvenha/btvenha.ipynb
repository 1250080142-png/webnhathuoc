```python

points = [
    (10.7626, 106.6602), # Thành phố Hồ Chí Minh
    (21.0285, 105.8542), # Hà Nội
    (16.0544, 108.2022), # Đà Nẵng
    (12.2451, 109.1943)  # Nha Trang
]


centroid_lat = sum([point[0] for point in points]) / len(points)
centroid_lon = sum([point[1] for point in points]) / len(points)


centroid_coords = [centroid_lat, centroid_lon]


centroid_dict = {
    "vĩ độ": centroid_lat,
    "kinh độ": centroid_lon
}


print(f"Danh sách các điểm: {points}")
print(f"Tâm của các điểm là: {centroid_coords}")
print(f"Từ điển lưu trữ tâm: {centroid_dict}")
```
