```python
input_file = "coordinate.text"
output_file = "output_coordinates.text"
try:
    with open(input_file, "r") as infile:
        coordinates = infile.readlines()
    with open(output_file, "w") as outfile:
        for line in coordinates:
            lat, lon = line.strip().split(",")
            outfile.write(f"Latitude: {lat}, Longitude: {lon}\n")
    print(f"Tọa độ đã được ghi vào file {output_file}")
except FileNotFoundError:
    print("Lỗi: Tập tin {input_file} đã không tìm thấy.")
```

```python
def parse_coordinates(line):
    try:
        lat,lon = line.strip().split(",")
        lat = float(lat)
        lon = float(lon)
        return lat, lon
    except ValueError as e:
        print(f"Error: {e}. Could not parse line: {line.strip()}")
        return None
    except Exception as e:
        print(f"An unexcepted error occrured: {e}")
        return None
```

```python
line = "invalid data"
coordinates = parse_coordinates(line)
if coordinates:
    print(f"Parse coordinates: {coordinates}")
```

```python
def process_geospatail_file(input_file):
    try:
        with open (input_file, "r") as infile:
            for liner in infile:
                coordinates = parse_coordinates(line)
                if coordinates:
                    print(f"Process coordinates: {coordinates}")
    except FileNotFoundError:
        print(f"Error: The file {input_file} was not found")
    except Exception as e:
        print(f"An unexcept error occurred while processing this file: {e}")
    finally:
        print(f"Finished processing: {input_file}")
```

```python
file_path = "coordinate.text"
process_geospatail_file(file_path)
```

```python
import requests
url=("https://d2ad6b4ur7yvpq.cloudfront.net/naturalearth-3.3.0/ne_110m_populated_places.geojson")
def fetch_geojson(url):
    try:
        response = requests.get(url)
        response.raise_for_status()
        geojson_data = response.json()
        return geojson_data
    except requests.exceptions.HTTPError as http_err:
        print(f"HTTP error occurred: {http_err}")
    except requests.exceptions.ConnectionError as conn_err:
        print(f"Error connecting to the server: {conn_err}")
    except Exception as err:
        print(f"An error occurred: {err}")
    return None
```

```python
geojson_data = fetch_geojson(url)

if geojson_data:
    features = geojson_data.get("features", [])
    print(f"Số thành phố: {len(features)}")

    for feature in features[:5]:
        city_name = feature["properties"].get("name")
        country_name = feature["properties"].get("country")
        coordinates = feature["geometry"]["coordinates"]
        print("Tên: {city_name}, Quốc gia: {country_name}, Tọa độ: {coordinates}")
```
