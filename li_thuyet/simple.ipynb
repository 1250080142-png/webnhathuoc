```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
```

```python
df = pd.DataFrame({
    'x' : [3.8 , 3.1, 2.9, 4.9, 5.1, 3.2, 5.1],
    'y' : [16, 13, 12, 20, 23, 14, 19]
})
df
```

```python
x = df[['x']]
y = df['y']
x.shape, y.shape
```

```python
lg = LinearRegression()
lg.fit(x, y)
lg.coef_, lg.intercept_
```

```python
y_pred = lg.predict(x)
y_pred
```

```python
mean_squared_error(y, y_pred)
```

```python
r2_score(y, y_pred)
```
