```python
import math

def calculate_distance(lat1, lon1, lat2, lon2):
    # Sử dụng công thức Haversine để tính khoảng cách thực tế (km)
    R = 6371  # Bán kính Trái Đất (km)
    
    phi1, phi2 = math.radians(lat1), math.radians(lat2)
    dphi = math.radians(lat2 - lat1)
    dlambda = math.radians(lon2 - lon1)
    
    a = math.sin(dphi / 2)**2 + \
        math.cos(phi1) * math.cos(phi2) * math.sin(dlambda / 2)**2
    
    return 2 * R * math.atan2(math.sqrt(a), math.sqrt(1 - a))

# Dữ liệu các chi nhánh Long Châu tại Tân Bình
coordinates = [
    {"name": "Long Châu 112-114 Trường Chinh", "lat": 10.7943523, "lon": 106.6510011},
    {"name": "Long Châu 9 (1010 Âu Cơ)", "lat": 10.7992779, "lon": 106.6374047},
    {"name": "Long Châu 47 Hoàng Hoa Thám", "lat": 10.7998922, "lon": 106.6471374},
    {"name": "Long Châu 797 CMT8", "lat": 10.7883638, "lon": 106.6617026},
    {"name": "Long Châu 1KA Bình Giã", "lat": 10.798831, "lon": 106.6441812}
]

# Điểm mốc để tính (Ví dụ: Ngã tư Bảy Hiền - 10.7938, 106.6545)
reference_point = (10.7938, 106.6545)

print(f"Khoảng cách từ điểm mốc {reference_point} đến các chi nhánh:")
for shop in coordinates:
    dist = calculate_distance(reference_point[0], reference_point[1], shop["lat"], shop["lon"])
    print(f"- {shop['name']}: {dist:.2f} km")
```
